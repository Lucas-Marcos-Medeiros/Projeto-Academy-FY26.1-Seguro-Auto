## Ajustes nos dados

Antes de começar a alterar algumas coisas no conjunto de dados, vamos realizar uma cópia deles com o nome 'casco':

```python
# Cópia de trabalho para não mexer no df original
casco = df.copy()

# Renomear colunas para snake_case (mantendo sentido do dicionário)
casco = casco.rename(columns={
    "COD_TARIF": "categoria",
    "REGIAO": "regiao",
    "COD_MODELO": "modelo",
    "ANO_MODELO": "ano",
    "SEXO": "sexo",
    "IDADE": "faixa_etaria",
    "EXPOSICAO1": "exposicao1",
    "PREMIO1": "premio1",
    "EXPOSICAO2": "exposicao2",   # não utilizado (segundo o dicionário)
    "PREMIO2": "premio2",         # não utilizado
    "IS_MEDIA": "is_media",
    "FREQ_SIN1": "freq_sin1",
    "INDENIZ1": "indeniz1",
    "FREQ_SIN2": "freq_sin2",
    "INDENIZ2": "indeniz2",
    "FREQ_SIN3": "freq_sin3",
    "INDENIZ3": "indeniz3",
    "FREQ_SIN4": "freq_sin4",
    "INDENIZ4": "indeniz4",
    "FREQ_SIN9": "freq_sin9",
    "INDENIZ9": "indeniz9",
    "ENVIO": "envio"
})

casco.head(3)
```

Verificar se há dados faltantes:

## Uso de dicionários para substituir códigos por categorias

* Para que possamos substituir os códigos de algumas variáveis pelas suas ocorrências, vamos criar dicionários.

* O dicionário é um tipo de dado em que temos o par 'chave: valor', sendo que a chave tem que ser única para aquele dicionário.

* Para realizar as substituições que vou mostrar, é preciso baixar os dados referentes ao segundo semestre de 2020 e lá estarão algumas planilhas importantes.

```python
casco.head()
```

### CATEGORIA

Para a variável 'CATEGORIA':

Os códigos estão na tabela auxiliar 'auto_cat' que possui definição no Manual da Autoseg;

Definição: auto_cat – código de descrição de categorias tarifárias;

Como são poucos códigos, vamos criar o dicionário de forma manual;

Realizar a substituição.

Manual da Autoseg:

Categoria:
* 1: Carro nacional - Passeio nacional
* 2: Carro importado - Passeio importado
* 3: Pick-up (nacional e importado)
* 4: Veículo de Carga (nacional e importado)
* 5: Motocicleta (nacional e importado)
* 6: Ônibus (nacional e importado)
* 7: Utilitários (nacional e importado)
* 9: Outros

```python
# Observação dos valores únicos da coluna categoria
casco['categoria'].unique()
```

```python
# criação do dicionário de categorias
cat_d = {
    1: 'Carro_nacional',
    2: 'Carro_importado',
    3: 'Pick-up',
    4: 'Veículo_carga',
    5: 'Motocicleta',
    6: 'Ônibus',
    7: 'Utilitários',
    9: 'Outros'
}

# Criamos uma nova coluna com a descrição da categoria (categoria_desc)
# e colocamos a descrição do dicionário usando a função map
casco['categoria_desc'] = casco['categoria'].map(cat_d).astype('string')

# (opcional) transformar em categórico para análises/plots mais leves
casco['categoria_desc'] = casco['categoria_desc'].astype('category')

# checar
casco[['categoria','categoria_desc']].head()
```

```python
casco.head(5)
```

### REGIÃO

Para a variável 'REGIÃO':

Os códigos estão na tabela auxiliar 'auto_reg' que possui definição no Manual da Autoseg;

Definição: auto_reg – código e descrição de regiões de circulação;

Como são poucos códigos, vamos criar o dicionário de forma manual;

Realizar a substituição.

```python
# eu especifiquei as subpastas onde os arquivos se encontram
# dentro do meu google drive: /dados/Autoseg2021A/
reg = pd.read_csv('/content/drive/My Drive/dados/SUSEP/RAW/auto_reg.csv', sep=';', encoding='latin1')
```

```python
# criar o dicionário usando a coluna 'CODIGO' como chave
# e a coluna 'DESCRICAO' para fazer a substituição
reg_d = reg.set_index('CODIGO')['DESCRICAO'].to_dict()
```

```python
reg_d
```

```python
casco.head()
```

```python
# 1) Garanta que a coluna 'regiao' está como inteiro "nullable" (Int64) para casar com as chaves do dict
casco["regiao_cod"] = pd.to_numeric(casco["regiao"], errors="coerce").astype("Int64")

# 2) Faça o map usando o dicionário (chaves inteiras -> descrição)
casco["regiao_desc"] = casco["regiao_cod"].map(reg_d)

# 3) (opcional) Transforme em categórico para melhorar performance em análises
# casco["regiao_desc"] = casco["regiao_desc"].astype("category")

# 4) Verifique
casco[["regiao_cod","regiao_desc"]].head()
casco["regiao_desc"].value_counts(dropna=False).head(10)
```

```python
casco["regiao_cod"].unique()
```

```python
casco.head(5)
```

```python
# Dropando as colunas 'categoria' e 'regiao' do nosso DF.
casco.drop(['categoria', 'regiao'], axis=1, inplace=True)
```

### MODELO

Para a variável 'MODELO':

* Há uma planilha pronta com os códigos e as descrições ('auto2_vei.csv'), vamos carregá-la;
* Criar o dicionário de substituição a partir da planilha;
* Realizar a substituição usando map() e não replace() para agilizar;
* Alterar o nome da variável para 'MODELO'.

```python
# checar quantas ocorrências de cada COD_MODELO existem
casco['modelo'].value_counts()
```

```python
# carregar a planilha com as codificações dos modelos de automóveis
vei = pd.read_csv('/content/drive/My Drive/dados/SUSEP/RAW/auto2_vei.csv', sep=';', encoding='latin1')
```

```python
vei.head()
```

```python
# criar o dicionário usando a coluna 'CODIGO' como chave
# e a coluna 'GRUPO' para fazer a substituição
# poderia ser a coluna 'GRUPO'
vei_d = vei.set_index('CODIGO')['GRUPO'].to_dict()
```

```python
# substituir os códigos pelas descrições usando map()
# já criando a nova variável
casco['modelo']= casco['modelo'].map(vei_d)
```

```python
# checar quantas ocorrências de cada valor existem após a alteração
casco['modelo'].value_counts().head(10)
```

```python
casco['modelo'].isna().sum()
```

```python
# Contar strings vazias ("") e espaços (" ")
print("Valores vazios (''):", (casco['modelo'] == '').sum())
print("Valores com espaço (' '):", (casco['modelo'] == ' ').sum())

# E também aqueles que podem ter mais de um espaço
print("Valores apenas com espaços:", casco['modelo'].str.strip().eq('').sum())
```

```python
# Mostrar algumas linhas onde o valor é nulo ou vazio
casco[casco['modelo'].isna() | casco['modelo'].str.strip().eq('')].head(10)
```

```python
# Substituição dos valores Nan por 'Não informado'
casco['modelo'] = casco['modelo'].replace(['', ' ', '  '], np.nan)
casco['modelo'] = casco['modelo'].fillna('Não informado')
```

### FAIXA ETARIA

Para a variável 'FAIXA ETÁRIA':

Os códigos estão na tabela auxiliar 'auto_idade' que possui definição no Manual da Autoseg;

Definição: auto_idade – código e descrição de faixas etárias;

Como são poucos códigos, vamos criar o dicionário de forma manual;

Realizar a substituição.

Manual da Autoseg:

Categoria:
* 0: Não informada
* 1: Entre 18 e 25 anos  
* 2: Entre 26 e 35 anos
* 3: Entre 36 e 45 anos
* 4: Entre 46 e 55 anos
* 5: Maior que 55 anos

```python
# Observação dos valores únicos da coluna faixa_etaria
casco['faixa_etaria'].unique()
```

```python
casco['faixa_etaria'].value_counts()
```

```python
# criação do dicionário de faixas_etárias
faixa_d = {
    0: 'Não informada',
    1: 'Entre 18 e 25 anos',
    2: 'Entre 26 e 35 anos',
    3: 'Entre 36 e 45 anos',
    4: 'Entre 46 e 55 anos',
    5: 'Maior que 55 anos',
}

# Criamos uma nova coluna com a descrição da categoria (categoria_desc)
# e colocamos a descrição do dicionário usando a função map
casco['faixa_desc'] = casco['faixa_etaria'].map(faixa_d).astype('string')

# (opcional) transformar em categórico para análises/plots mais leves
# casco['categoria_desc'] = casco['categoria_desc'].astype('category')

# checar
casco[['faixa_etaria','faixa_desc']].head()
```

```python
casco.head(5)
```

```python
# Dropando as colunas 'faixa_etaria' do nosso DF.
casco.drop(['faixa_etaria'], axis=1, inplace=True)
```

```python
casco.head()
```

```python
casco.columns
```

```python
casco.dtypes
```

```python
# Remover espaços em branco e substituir pontos por nada
casco['ano'] = casco['ano'].astype(str).str.strip().str.replace('.', '', regex=False)
```

```python
# Substituir valores problemáticos por NaN
casco['ano'] = casco['ano'].replace(['', ' ', '0000', '.', '   .', '    '], np.nan)

# Converter para numérico (forçando erro → NaN)
casco['ano'] = pd.to_numeric(casco['ano'], errors='coerce')
```

```python
# Converter para tipo inteiro compatível com NaN
casco['ano'] = casco['ano'].astype('Int64')
```

```python
casco['ano'].dtypes
casco['ano'].unique()
```

```python
# Substituir o ano 0 por <NA>
casco.loc[casco['ano'] == 0, 'ano'] = pd.NA
```

```python
# Corrigir anos com erros de digitação

# Corrigir o ano 201 -> 2001
casco.loc[casco['ano'] == 201, 'ano'] = 2001

# Corrigir anos acima de 2025 (ex: 2087, 2099, 2095...) subtraindo 100
casco.loc[casco['ano'] > 2025, 'ano'] = casco.loc[casco['ano'] > 2025, 'ano'] - 100

# Reconfirmar o tipo
casco['ano'] = casco['ano'].astype('Int64')
```
